## Pemeriksaan Kualitas Data

### 3.1 Pemeriksaan Impossible Values

*Impossible values* adalah nilai yang secara logis tidak mungkin ada, bukan hanya sekadar ekstrem. Contohnya: usia negatif, pendapatan negatif, atau persentase di atas 100%. Pemeriksaan ini penting karena nilai seperti ini tidak dapat dikoreksi dengan capping dan harus dihapus atau diselidiki secara manual.

Pemeriksaan dilakukan setelah seluruh proses scaling dan capping selesai, karena kolom-kolom seperti `age_scaled` dan `monthly_bill_capped` baru akan tersedia pada tahap tersebut. Pada tahap awal ini, kita lakukan pemeriksaan terhadap kolom-kolom numerik asli.


````python
# Pemeriksaan impossible values pada kolom numerik asli
checks = {
    'age (< 0 atau > 120)': ((df['age'] < 0) | (df['age'] > 120)).sum(),
    'monthly_income (< 0)': (df['monthly_income'] < 0).sum(),
    'monthly_bill (< 0)': (df['monthly_bill'] < 0).sum(),
    'internet_usage_gb (< 0)': (df['internet_usage_gb'] < 0).sum(),
    'call_minutes (< 0)': (df['call_minutes'] < 0).sum(),
    'support_tickets (< 0)': (df['support_tickets'] < 0).sum(),
    'churn (bukan 0 atau 1)': (~df['churn'].isin([0, 1])).sum(),
}

print("Pemeriksaan Impossible Values:")
print("-" * 45)
semua_ok = True
for kondisi, jumlah in checks.items():
    status = "OK" if jumlah == 0 else f"DITEMUKAN {jumlah} kasus"
    print(f"  {kondisi:<35}: {status}")
    if jumlah > 0:
        semua_ok = False

print("-" * 45)
if semua_ok:
    print("Tidak ditemukan impossible values pada dataset.")
else:
    print("Terdapat impossible values yang perlu ditangani.")
````

### 3.2 Penanganan Data Duplikat

Data duplikat adalah baris yang memiliki nilai identik pada seluruh kolom. Keberadaan baris duplikat dapat menyebabkan model machine learning menjadi bias karena pola tertentu akan terhitung lebih dari sekali dalam proses pelatihan. Penyebab umum duplikasi antara lain: proses join tabel yang tidak tepat, pengumpulan data dari berbagai sumber yang tumpang tindih, atau kesalahan entri data.

Fungsi `df.duplicated()` mengembalikan Series boolean, di mana `True` menandai baris yang merupakan duplikat dari baris sebelumnya. Fungsi `.sum()` kemudian menghitung total baris yang terduplikasi.


````python
n_dup = df.duplicated().sum()
print(f"Jumlah baris duplikat: {n_dup}")

if n_dup > 0:
    df.drop_duplicates(inplace=True)
    print(f"{n_dup} baris duplikat berhasil dihapus.")
    print(f"Shape setelah pembersihan: {df.shape}")
else:
    print("Tidak ditemukan baris duplikat. Tidak ada tindakan yang diperlukan.")

print(f"Jumlah baris setelah pengecekan duplikat: {len(df)}")
````

### 3.3 Penanganan Missing Values

Missing values atau nilai yang hilang adalah salah satu masalah paling umum dalam dataset nyata. Keberadaannya dapat menyebabkan error pada banyak algoritma machine learning yang tidak dapat memproses nilai kosong (NaN). Strategi penanganan meliputi: imputasi (mengisi dengan nilai tertentu seperti mean, median, atau modus), penghapusan baris/kolom, atau menggunakan algoritma yang toleran terhadap missing values.

Fungsi `df.isnull()` mengembalikan DataFrame boolean yang berisi `True` di posisi nilai yang hilang. Fungsi `.sum()` menjumlahkan total nilai hilang per kolom.


````python
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_df = pd.DataFrame({
    'Jumlah Missing': missing,
    'Persentase (%)': missing_pct
})

print("Pemeriksaan Missing Values per Kolom:")
print("-" * 45)
if missing_df['Jumlah Missing'].sum() == 0:
    print("Tidak ditemukan missing values pada seluruh kolom dataset.")
    print("Dataset dalam kondisi lengkap dan tidak memerlukan imputasi.")
else:
    display(missing_df[missing_df['Jumlah Missing'] > 0].sort_values('Jumlah Missing', ascending=False))
````

#### Fitur `tenure_months`

**Tujuan:** Mengukur berapa lama seorang pelanggan telah berlangganan layanan, dihitung dalam satuan bulan.

**Alasan pembuatan fitur:** Kolom `signup_date` dalam bentuk tanggal mentah tidak dapat langsung digunakan oleh model machine learning. Namun, informasi yang terkandung di dalamnya sangat berharga: pelanggan yang baru bergabung secara historis lebih rentan churn dibanding pelanggan lama yang sudah loyal. Dengan mengkonversinya menjadi `tenure_months`, kita mengekstrak informasi temporal yang bermakna dan dapat dipelajari oleh model.

**Rumus:**

```
tenure_months = (tanggal_referensi - signup_date).hari / 30
```

Tanggal referensi yang digunakan adalah **25 Juni 2026** (tanggal pengerjaan proyek). Pembagian dengan 30 mengkonversi satuan dari hari ke bulan, kemudian dibulatkan ke bawah menggunakan operator `//` (floor division).

**Contoh:** Jika seorang pelanggan mendaftar pada 1 Januari 2019, maka:
- Selisih hari dari 25 Juni 2026 = sekitar 2.732 hari
- tenure_months = 2.732 // 30 = **91 bulan**

Berdasarkan data, `tenure_months` berkisar antara **51 hingga 91 bulan**, dengan rata-rata **70,6 bulan**. Setelah fitur baru dibuat, kolom `signup_date` dihapus karena informasinya sudah terekstrak.



#### Fitur `financial_risk`

**Tujuan:** Menandai pelanggan yang tagihan bulanannya melebihi pendapatan bulanannya, sebagai indikator tekanan finansial.

**Alasan pembuatan fitur:** Secara logis, pelanggan yang membayar tagihan lebih besar dari pendapatannya berada dalam tekanan finansial. Kondisi ini meningkatkan probabilitas pelanggan tersebut untuk churn karena ketidakmampuan membayar atau mencari layanan yang lebih murah.

**Rumus:**

```
financial_risk = 1  jika monthly_bill > monthly_income
financial_risk = 0  jika monthly_bill <= monthly_income
```

Implementasinya menggunakan `np.where()`, yaitu fungsi numpy yang bekerja seperti IF-THEN-ELSE secara vektorisasi (berlaku pada seluruh baris sekaligus tanpa perlu perulangan).

Berdasarkan data, terdapat **23 pelanggan** (1,9%) yang teridentifikasi berisiko finansial.


````python
# Fitur tenure_months
if 'signup_date' in df.columns:
    df['signup_date'] = pd.to_datetime(df['signup_date'])
    tanggal_referensi = pd.to_datetime('2026-06-25')
    df['tenure_months'] = ((tanggal_referensi - df['signup_date']).dt.days // 30).astype(int)
    df.drop(columns=['signup_date'], inplace=True)
    print("Fitur 'tenure_months' berhasil dibuat dan kolom 'signup_date' dihapus.")
    print(f"  Rentang tenure : {df['tenure_months'].min()} - {df['tenure_months'].max()} bulan")
    print(f"  Rata-rata      : {df['tenure_months'].mean():.1f} bulan")

# Fitur financial_risk
if 'monthly_bill' in df.columns and 'monthly_income' in df.columns:
    df['financial_risk'] = np.where(df['monthly_bill'] > df['monthly_income'], 1, 0)
    n_risk = df['financial_risk'].sum()
    print(f"\nFitur 'financial_risk' berhasil dibuat.")
    print(f"  Pelanggan berisiko finansial : {n_risk} ({n_risk/len(df)*100:.1f}%)")

print("\n5 baris pertama setelah pembuatan fitur baru:")
display(df.head())
````

### 4.3 Identifikasi Anomali Perilaku Pelanggan

Fitur anomali adalah fitur biner yang menandai pelanggan yang menunjukkan perilaku tidak biasa atau tidak sesuai pola umum. Fitur seperti ini sering kali menjadi sinyal penting bagi model karena mencerminkan kondisi-kondisi spesifik yang sulit ditangkap oleh fitur tunggal.

Dibuat tiga indikator anomali:

**- Anomali 1: `anomaly_pay_not_use` - Pay-Not-Use**

**Definisi:** Pelanggan yang membayar tagihan di atas median namun penggunaan internetnya nol.

**Alasan:** Pelanggan yang membayar mahal tetapi tidak menggunakan layanan internet menunjukkan potensi ketidakpuasan atau ketidaktahuan tentang paket mereka. Kondisi ini sering mendahului keputusan untuk churn karena pelanggan merasa tidak mendapat nilai dari uang yang dibayarkan.

**Rumus:**
```
anomaly_pay_not_use = 1  jika monthly_bill > median(monthly_bill) DAN internet_usage_gb == 0
anomaly_pay_not_use = 0  selain itu
```

Catatan: Berdasarkan data aktual, tidak ada pelanggan dengan `internet_usage_gb == 0` (nilai minimum adalah 1,1 GB), sehingga anomali ini bernilai 0 untuk semua baris. Fitur tetap dibuat sebagai antisipasi jika dataset diperbarui.

**- Anomali 2: `anomaly_happy_leaver` - Happy Leaver**

**Definisi:** Pelanggan yang tidak pernah mengajukan tiket dukungan (`support_tickets == 0`) namun tercatat sebagai churned.

**Alasan:** Secara umum, pelanggan yang tidak puas akan mengajukan keluhan sebelum memutuskan pergi. Namun, ada segmen pelanggan yang pergi tanpa mengeluh — mereka mungkin berpindah ke kompetitor karena faktor harga atau promosi yang lebih menarik, bukan karena buruknya layanan. Mengidentifikasi segmen ini penting karena strategi retensi yang tepat untuk mereka berbeda (fokus pada harga dan loyalitas, bukan kualitas layanan).

**Rumus:**
```
anomaly_happy_leaver = 1  jika support_tickets == 0 DAN churn == 1
anomaly_happy_leaver = 0  selain itu
```

Berdasarkan data, terdapat **21 pelanggan** yang masuk kategori happy leaver.

**- Anomali 3: `anomaly_genz_call_enthusiast` - GenZ Call Enthusiast**

**Definisi:** Pelanggan berusia 27 tahun ke bawah yang durasi panggilannya di atas median.

**Alasan:** Secara demografis, generasi muda (Gen Z, lahir sekitar 1997–2012) cenderung lebih banyak menggunakan layanan data dan pesan teks dibanding panggilan suara. Pelanggan muda dengan durasi panggilan tinggi merupakan pola yang tidak biasa dan bisa menjadi sinyal perilaku khusus yang relevan untuk segmentasi atau strategi layanan.

**Rumus:**
```
anomaly_genz_call_enthusiast = 1  jika age <= 27 DAN call_minutes > median(call_minutes)
anomaly_genz_call_enthusiast = 0  selain itu
```

Berdasarkan data, terdapat **121 pelanggan** yang masuk kategori ini.


````python
# Anomali 1: Pay-Not-Use
if 'internet_usage_gb' in df.columns and 'monthly_bill' in df.columns:
    df['anomaly_pay_not_use'] = np.where(
        (df['monthly_bill'] > df['monthly_bill'].median()) & (df['internet_usage_gb'] == 0),
        1, 0
    )
else:
    df['anomaly_pay_not_use'] = 0

# Anomali 2: Happy Leaver
if 'support_tickets' in df.columns and 'churn' in df.columns:
    df['anomaly_happy_leaver'] = np.where(
        (df['support_tickets'] == 0) & (df['churn'] == 1),
        1, 0
    )
else:
    df['anomaly_happy_leaver'] = 0

# Anomali 3: GenZ Call Enthusiast
if 'age' in df.columns and 'call_minutes' in df.columns:
    df['anomaly_genz_call_enthusiast'] = np.where(
        (df['age'] <= 27) & (df['call_minutes'] > df['call_minutes'].median()),
        1, 0
    )
else:
    df['anomaly_genz_call_enthusiast'] = 0

print("Pembuatan kolom indikator anomali selesai.")
print()
for col in ['anomaly_pay_not_use', 'anomaly_happy_leaver', 'anomaly_genz_call_enthusiast']:
    n = df[col].sum()
    print(f"  {col}: {n} pelanggan ({n/len(df)*100:.1f}%)")

print()
print("Contoh 15 baris pertama untuk kolom anomali:")
print(df[['anomaly_pay_not_use', 'anomaly_happy_leaver', 'anomaly_genz_call_enthusiast']].head(15))
````

### 5.2 Implementasi Capping dan Scaling

#### 5.2.1 Capping Outlier Tambahan (IQR) untuk Kolom `monthly_bill`, `internet_usage_gb`, `call_minutes`, `support_tickets`

Alasan menggunakan capping adalah karena outlier pada dataset ini merupakan **outlier valid** yang masih merepresentasikan kondisi nyata, digunakan teknik **Capping (Winsorization)** untuk membatasi nilai ekstrem tanpa menghapus data. Pendekatan ini dipilih agar:
- Mempertahankan seluruh data dan informasi penting.
- Mengurangi pengaruh nilai ekstrem terhadap proses analisis dan pemodelan.
- Menjaga distribusi data tetap representatif terhadap kondisi sebenarnya.
Metode IQR (*Interquartile Range*) adalah standar statistik yang paling umum digunakan untuk deteksi outlier. Rumus batas capping:

```
Q1   = kuartil bawah (25% data terendah)
Q3   = kuartil atas (25% data tertinggi)
IQR  = Q3 - Q1

Batas bawah = Q1 - 1.5 × IQR
Batas atas  = Q3 + 1.5 × IQR
```

Nilai yang lebih kecil dari batas bawah akan diubah menjadi batas bawah, dan nilai yang lebih besar dari batas atas akan diubah menjadi batas atas. Fungsi `.clip(lower, upper)` di pandas melakukan ini secara efisien.

Hasilnya disimpan dalam kolom baru dengan akhiran `_capped` agar kolom asli tetap tersimpan sebagai referensi.


````python
# Fungsi untuk menerapkan capping outlier dengan IQR
def cap_outliers_iqr(df_col):
    Q1 = df_col.quantile(0.25)
    Q3 = df_col.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return df_col.clip(lower=lower_bound, upper=upper_bound)

# Kolom-kolom yang akan ditangani dengan capping
columns_to_cap = ['monthly_bill', 'internet_usage_gb', 'call_minutes', 'support_tickets']

for col in columns_to_cap:
    if col in df.columns:
        df[f'{col}_capped'] = cap_outliers_iqr(df[col])
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        upper = Q3 + 1.5 * IQR
        print(f"Outlier '{col}' berhasil di-capping. Batas atas: {upper:.1f} (max asli: {df[col].max():.1f})")

print("\nRingkasan Statistik Setelah Capping:")
display(df[columns_to_cap + [f'{c}_capped' for c in columns_to_cap]].describe())
````

#### 5.2.2 Deteksi Outlier (IQR) dan Scaling Min-Max untuk `monthly_income` dan `age`

Untuk dua kolom ini, selain capping juga dilakukan **scaling** menggunakan metode Min-Max Normalization. Berikut penjelasan mengapa scaling diperlukan dan mengapa Min-Max dipilih:

Algoritma machine learning seperti Regresi Logistik, Support Vector Machine (SVM), dan K-Nearest Neighbors (KNN) sangat sensitif terhadap perbedaan skala antar fitur. Jika satu fitur bernilai ratusan ribu sementara fitur lain hanya bernilai 0-1, maka fitur dengan nilai besar akan mendominasi perhitungan jarak atau gradien, membuat fitur lain menjadi tidak berpengaruh meskipun secara informasi sama pentingnya.

**Alasan pemilihan Min-Max untuk dataset ini:**
- Rentang domain yang jelas dapat ditentukan: usia manusia antara 18-80 tahun, dan pendapatan bulanan dalam konteks dataset ini secara wajar berada di antara 5.000-100.000.
- Min-Max cocok untuk algoritma berbasis pohon (Decision Tree, Random Forest) dan neural network sederhana yang akan digunakan dalam pemodelan selanjutnya.
- Setelah capping, nilai ekstrem sudah dibatasi, sehingga masalah Min-Max yang sensitif terhadap outlier sudah diminimalkan.

**Rumus Min-Max Normalization:**

```
x_scaled = (x - x_min_domain) / (x_max_domain - x_min_domain)

Untuk age          : x_min = 18, x_max = 80
Untuk monthly_income: x_min = 5.000, x_max = 100.000
```

Hasil di-clip ke [0, 1] untuk memastikan nilai yang sudah dicap IQR tidak keluar dari rentang target.


````python
# 1. PENANGANAN OUTLIER & SCALING UNTUK MONTHLY INCOME
if 'monthly_income' in df.columns:
    Q1_inc = df['monthly_income'].quantile(0.25)
    Q3_inc = df['monthly_income'].quantile(0.75)
    IQR_inc = Q3_inc - Q1_inc
    batas_bawah_inc = Q1_inc - 1.5 * IQR_inc
    batas_atas_inc = Q3_inc + 1.5 * IQR_inc

    df['monthly_income_cleaned'] = df['monthly_income'].clip(lower=batas_bawah_inc, upper=batas_atas_inc)

    # Scaling Min-Max dengan rentang domain 5.000 - 100.000
    df['monthly_income_scaled'] = (df['monthly_income_cleaned'] - 5000) / (100000 - 5000)
    df['monthly_income_scaled'] = df['monthly_income_scaled'].clip(0, 1)

    df.drop(columns=['monthly_income_cleaned'], inplace=True)
    print("Deteksi outlier dan Scaling Min-Max untuk 'monthly_income' selesai.")
    print(f"  Batas atas IQR  : {batas_atas_inc:.1f}")
    print(f"  Rentang scaled  : [{df['monthly_income_scaled'].min():.3f}, {df['monthly_income_scaled'].max():.3f}]")

# 2. PENANGANAN OUTLIER & SCALING UNTUK AGE
if 'age' in df.columns:
    Q1_age = df['age'].quantile(0.25)
    Q3_age = df['age'].quantile(0.75)
    IQR_age = Q3_age - Q1_age
    batas_bawah_age = Q1_age - 1.5 * IQR_age
    batas_atas_age = Q3_age + 1.5 * IQR_age

    df['age_cleaned'] = df['age'].clip(lower=batas_bawah_age, upper=batas_atas_age)

    # Scaling Min-Max dengan rentang domain 18 - 80
    df['age_scaled'] = (df['age_cleaned'] - 18) / (80 - 18)
    df['age_scaled'] = df['age_scaled'].clip(0, 1)

    df.drop(columns=['age_cleaned'], inplace=True)
    print("\nDeteksi outlier dan Scaling Min-Max untuk 'age' selesai.")
    print(f"  Rentang usia asli: [{df['age'].min()}, {df['age'].max()}]")
    print(f"  Rentang scaled   : [{df['age_scaled'].min():.3f}, {df['age_scaled'].max():.3f}]")

    print("\nContoh 15 baris pertama untuk monthly_income dan age setelah scaling:")
    print(df[['monthly_income', 'monthly_income_scaled', 'age', 'age_scaled']].head(15))
````

## Transformasi Data Kategorikal dan Data Teks

### 6.1 Pembersihan Whitespace pada Kolom Kategorikal

Whitespace adalah karakter spasi, tab, atau newline yang muncul di awal atau akhir string. Dalam data yang dikumpulkan dari berbagai sumber atau melalui entri manual, whitespace sering kali tidak terlihat namun menyebabkan ketidakkonsistenan kategori.

Contoh masalah: `'Male'` dan `'Male '` (dengan spasi di akhir) akan dianggap sebagai dua kategori berbeda oleh Python, meskipun secara makna sama. Ini akan menyebabkan encoding yang salah dan meningkatkan jumlah kategori secara tidak perlu.

Fungsi `.str.strip()` menghapus whitespace di awal dan akhir setiap string. Proses ini diterapkan pada semua kolom dengan tipe data `object` (string).


````python
# Membersihkan spasi di awal dan akhir string untuk kolom objek
text_columns = df.select_dtypes(include='object').columns

for col in text_columns:
    df[col] = df[col].astype(str).str.strip()

print(f"Karakter whitespace pada {len(text_columns)} kolom bertipe string telah dibersihkan.")
print(f"Kolom yang dibersihkan: {list(text_columns)}")
````

### 6.2 Encoding Variabel Kategorikal

Algoritma machine learning pada umumnya hanya dapat memproses data numerik. Oleh karena itu, seluruh kolom kategorikal (string) harus dikonversi menjadi representasi numerik. Pemilihan metode encoding bergantung pada sifat variabel kategorikal tersebut.


#### 6.2.1 Encoding Ordinal dan Biner untuk `gender`, `education_level`, `contract_type`, dan `employment_status`

**Encoding Biner untuk `gender`:**

Gender hanya memiliki dua nilai (`Male`, `Female`), sehingga cukup direpresentasikan dengan satu kolom biner: `Female = 0` dan `Male = 1`. Metode ini disebut **Binary Encoding** atau Label Encoding untuk variabel biner.

**Encoding Ordinal untuk `education_level`:**

Tingkat pendidikan memiliki urutan hierarki yang bermakna: Primary < Secondary < Bachelor < Master < PhD. Karena ada hubungan urutan yang jelas antar kategori, metode **Ordinal Encoding** tepat digunakan. Angka yang diberikan mencerminkan peringkat relatif, bukan jarak absolut.

Pemetaan: `Primary=0`, `Secondary=1`, `Bachelor=2`, `Master=3`, `PhD=4`

**Encoding Ordinal untuk `contract_type`:**

Jenis kontrak juga memiliki urutan yang bermakna dalam konteks komitmen pelanggan: Monthly (berlangganan bulanan, komitmen paling rendah) < 6-Month < 12-Month (komitmen paling tinggi).

Pemetaan: `Monthly=0`, `6-Month=1`, `12-Month=2`

**One-Hot Encoding untuk `employment_status`:**

Status pekerjaan (`Employed`, `Self-Employed`, `Student`, `Unemployed`, `Retired`) adalah variabel **nominal**, artinya tidak ada urutan hierarki yang bermakna antar kategori. Menggunakan ordinal encoding untuk variabel nominal akan memperkenalkan hubungan ordinal palsu yang tidak ada. Oleh karena itu, digunakan **One-Hot Encoding**: setiap kategori menjadi kolom biner tersendiri.

Fungsi `pd.get_dummies()` mengkonversi kolom kategorikal menjadi beberapa kolom biner, satu per kategori.


````python
# 1. Gender - binary encoding
if 'gender' in df.columns:
    df['gender'] = df['gender'].map({'Female': 0, 'Male': 1, 'female': 0, 'male': 1})
    print("Encoding 'gender': Female=0, Male=1")
    print(df['gender'].value_counts().to_string())

# 2. Education Level - ordinal encoding
edu_mapping = {'Primary': 0, 'Secondary': 1, 'Bachelor': 2, 'Master': 3, 'PhD': 4}
if 'education_level' in df.columns:
    df['education_level_encoded'] = df['education_level'].map(edu_mapping)
    print("\nEncoding 'education_level' (ordinal):")
    print(df['education_level_encoded'].value_counts().sort_index().to_string())

# 3. Contract Type - ordinal encoding
contract_mapping = {'Monthly': 0, '6-Month': 1, '12-Month': 2}
if 'contract_type' in df.columns:
    df['contract_type_encoded'] = df['contract_type'].map(contract_mapping)
    print("\nEncoding 'contract_type' (ordinal):")
    print(df['contract_type_encoded'].value_counts().sort_index().to_string())

# 4. Employment Status - one-hot encoding
if 'employment_status' in df.columns:
    df_emp_dummies = pd.get_dummies(df['employment_status'], prefix='emp', drop_first=False, dtype=int)
    df = pd.concat([df, df_emp_dummies], axis=1)
    print(f"\nOne-Hot Encoding 'employment_status': {list(df_emp_dummies.columns)}")

print("\nProses encoding selesai.")
display(df.head())
````

#### 6.2.2 One-Hot Encoding untuk Kolom `city`

Kolom `city` berisi 11 kota yang semuanya bersifat nominal (tidak ada urutan hierarki antar kota). Oleh karena itu, digunakan One-Hot Encoding yang menghasilkan 11 kolom biner baru, masing-masing mewakili satu kota.

Opsi `dtype=int` memastikan nilai yang dihasilkan adalah integer (0 atau 1), bukan boolean.


````python
# One-Hot Encoding untuk kolom 'city'
if 'city' in df.columns:
    df_city_dummies = pd.get_dummies(df['city'], prefix='city', dtype=int)
    df = pd.concat([df, df_city_dummies], axis=1)
    print(f"One-Hot Encoding 'city' berhasil: {df_city_dummies.shape[1]} kolom baru ditambahkan.")
    print(f"Kota-kota: {list(df_city_dummies.columns)}")

display(df.head())
````

### 6.3 Preprocessing dan Count Vectorizer untuk `customer_feedback`

Kolom `customer_feedback` berisi teks (kalimat panjang), yang tidak bisa langsung digunakan oleh model. Dilakukan dua pendekatan untuk mengekstrak informasi dari kolom ini:

#### Kategorisasi `customer_feedback`

Terdapat 10 template kalimat feedback yang masing-masing merepresentasikan satu topik keluhan atau sentimen. Pendekatan keyword matching digunakan untuk mengklasifikasikan setiap feedback ke dalam salah satu dari empat kategori numerik:

| Nilai | Kategori | Kata Kunci yang Cocok | Contoh Feedback |
|-------|----------|----------------------|-----------------|
| 0 | Masalah Biaya/Tagihan | `billing`, `expensive` | "Billing issues occurred multiple times", "Too expensive compared to competitors" |
| 1 | Masalah Jaringan/Sinyal | `poor`, `unreliable`, `drops`, `slow` | "Coverage is poor in my location", "Internet speed is very slow" |
| 2 | Layanan Customer Support | `support` | "Customer support was helpful" |
| 3 | Sentimen Positif/Puas | `happy`, `satisfied`, `good` | "Satisfied with the service", "Happy with data packages" |

#### Apply Count Vectorizer untuk Kata Kunci `customer_feedback`

Selain kategorisasi, dibuat juga fitur biner untuk setiap kata kunci penting menggunakan pendekatan Count Vectorizer yang disederhanakan. Setiap kata kunci menghasilkan satu kolom biner baru: bernilai 1 jika kata kunci tersebut muncul dalam feedback, dan 0 jika tidak. Ini memberikan representasi yang lebih granular dibanding kategori tunggal.

10 kata kunci yang digunakan: `billing`, `support`, `helpful`, `poor`, `expensive`, `unreliable`, `happy`, `satisfied`, `drops`, `slow`.


````python
if 'customer_feedback' in df.columns:
    # Kategorisasi berbasis keyword
    df['feedback_category'] = -1
    fb_lower = df['customer_feedback'].astype(str).str.lower()

    df.loc[fb_lower.str.contains('billing|expensive'), 'feedback_category'] = 0
    df.loc[fb_lower.str.contains('poor|unreliable|drops|slow'), 'feedback_category'] = 1
    df.loc[fb_lower.str.contains('support'), 'feedback_category'] = 2
    df.loc[fb_lower.str.contains('happy|satisfied|good'), 'feedback_category'] = 3

    print("Kategorisasi customer_feedback:")
    label_map = {0: 'Biaya/Tagihan', 1: 'Jaringan/Sinyal', 2: 'Customer Support', 3: 'Positif/Puas', -1: 'Tidak Terklasifikasi'}
    for k, v in df['feedback_category'].value_counts().sort_index().items():
        print(f"  {k} ({label_map[k]}): {v} pelanggan")

    # Count Vectorizer berbasis kata kunci
    keywords = ['billing', 'support', 'helpful', 'poor', 'expensive',
                'unreliable', 'happy', 'satisfied', 'drops', 'slow']

    for word in keywords:
        df[f'count_{word}'] = df['customer_feedback'].str.lower().str.contains(word, na=False).astype(int)

    print(f"\nCount Vectorizer berhasil: {len(keywords)} kolom keyword ditambahkan.")
    print("Contoh kolom hasil vectorizer:")
    print(df[[f'count_{w}' for w in keywords[:5]]].head(10))
````

Setiap histogram dilengkapi dengan garis Mean (merah) dan Median (hijau) untuk membantu memahami apakah distribusi simetris atau condong ke salah satu arah:
- Jika Mean > Median: distribusi masih condong ke kanan (right-skewed)
- Jika Mean < Median: distribusi condong ke kiri (left-skewed)
- Jika Mean = Median: distribusi mendekati simetris

### 8.7 Korelasi Antar Variabel

Heatmap korelasi menampilkan koefisien korelasi Pearson antara setiap pasang variabel numerik. Korelasi Pearson mengukur kekuatan dan arah hubungan linear antara dua variabel, dengan nilai berkisar antara -1 (korelasi negatif sempurna) hingga +1 (korelasi positif sempurna). Nilai mendekati 0 menandakan tidak ada hubungan linear.

**Heatmap Korelasi:**


````python
# Pilih kolom numerik utama (tidak termasuk one-hot dummies untuk keterbacaan)
core_cols = [c for c in df.columns
             if not c.startswith('city_') and not c.startswith('emp_') and not c.startswith('count_')]
corr_matrix = df[core_cols].corr(numeric_only=True)

plt.figure(figsize=(16, 12))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
sns.heatmap(corr_matrix,
            mask=mask,
            annot=True,
            fmt='.2f',
            cmap='coolwarm',
            center=0,
            linewidths=0.5,
            annot_kws={'size': 7},
            square=True)
plt.title('12.6 Heatmap Korelasi Antar Variabel Numerik (Segitiga Bawah)', fontweight='bold')
plt.tight_layout()
plt.show()
````

**Korelasi terhadap Churn:**

````python
churn_corr = df.corr(numeric_only=True)['churn'].drop('churn').sort_values(key=abs, ascending=False).head(15)

plt.figure(figsize=(10, 6))
colors_bar = ['#E74C3C' if v > 0 else '#3498DB' for v in churn_corr.values]
plt.barh(churn_corr.index[::-1], churn_corr.values[::-1],
         color=colors_bar[::-1], edgecolor='white', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.title('Top 15 Fitur dengan Korelasi Terkuat terhadap Churn', fontweight='bold')
plt.xlabel('Korelasi Pearson (merah = positif, biru = negatif)')
plt.grid(axis='x', alpha=0.4)
plt.tight_layout()
plt.show()

print("\nTop 15 Korelasi terhadap 'churn':")
for feat, corr in churn_corr.items():
    arah = "positif" if corr > 0 else "negatif"
    print(f"  {feat:<35}: {corr:+.3f} ({arah})")
````

**Interpretasi Korelasi:**

Korelasi Pearson mengukur hubungan *linear* antara dua variabel. Nilai korelasi terhadap `churn` mengindikasikan seberapa kuat suatu fitur berhubungan dengan kemungkinan churn:

- **Korelasi positif** (merah): Semakin besar nilai fitur, semakin tinggi kemungkinan churn.
- **Korelasi negatif** (biru): Semakin besar nilai fitur, semakin rendah kemungkinan churn.

Fitur-fitur dengan korelasi absolut tertinggi terhadap churn menjadi kandidat fitur prediktif yang paling kuat dan akan menjadi fokus analisis mendalam di notebook selanjutnya. Perlu diingat bahwa korelasi Pearson hanya mengukur hubungan linear — hubungan non-linear tidak akan terdeteksi di sini dan memerlukan teknik analisis yang lebih lanjut (seperti mutual information atau analisis berbasis model tree).
